```python
np.random.seed(0)
random.seed(0)

history = np.loadtxt(r"./training.txt", dtype=int).reshape(-1,3)
#Define your algorithm here or define your own functions and procedure
def rock_paper_scissor(history=None, last_choice=None):
    # define your function here
    beats = {0: 1, 1: 2, 2: 0} 

    if history is None or len(history) < 2:
        return random.randint(0, 2)

    # transition matrix from move history
    transitions = np.zeros((3, 3), dtype=int)
    for k in range(len(history) - 1):
        cpu_now  = history[k][0]
        cpu_next = history[k + 1][0]
        transitions[cpu_now][cpu_next] += 1

    # predict next move based on last move
    last_cpu_move = history[-1][0]
    row = transitions[last_cpu_move]

    if row.sum() == 0:
        return random.randint(0, 2)

    predicted_cpu_move = np.argmax(row)
    
    return beats[predicted_cpu_move]
```

```python
print(f'Algorithm says you should choose: {rock_paper_scissor(history)}')
```
